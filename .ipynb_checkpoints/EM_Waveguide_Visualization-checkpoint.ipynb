### Solución analítica - caso general

Considere una guía de onda rectangular de dimensiones $(a,b)$, donde $a > b$, con paredes conductoras. Suponiendo que las paredes conductoras están hechas de un conductor perfecto, el problema tendría las siguientes condiciones de frontera:

$$
\begin{aligned}
\mathbf{E}_{\parallel}&=0\\[6pt]
\mathbf{B}_{\perp}&=0
\end{aligned}
\tag{1}
$$

Además, considere que las ondas son monocromáticas y se propagan en dirección z, lo que permite interpretar el campo eléctrico $\tilde{E}$ y magnético $\tilde{B}$ como ondas planas:

$$
\begin{aligned}
\widetilde{\mathbf E}(x,y,z,t) &=\mathbf E_0(x,y)e^{i(k_z z-\omega t)}\\[6pt]
\widetilde{\mathbf B}(x,y,z,t) &=\mathbf B_0(x,y)e^{i(k_z z-\omega t)}
\end{aligned}
\tag{2}
$$

Estas ecuaciones deben cumplir tanto las condiciones de frontera (1) como las ecuaciones de Maxwell: 

$$
\begin{aligned}
\nabla\cdot\widetilde{\mathbf E}&=0\\[6pt]
\nabla\cdot\widetilde{\mathbf B}&=0\\[6pt]
\nabla\times\widetilde{\mathbf E}&=-\frac{\partial\widetilde{\mathbf B}}{\partial t}\\[6pt]
\nabla\times\widetilde{\mathbf B}&=\frac{1}{c^2}\frac{\partial\widetilde{\mathbf E}}{\partial t}
\end{aligned}
\tag{3}
$$

Para que se cumplan las condiciones y las ecuaciones, tanto $B_0$ como $E_0$ deben cumplirlas. Escribiendo ambos campos en sus componentes:

$$
\begin{aligned}
\mathbf E_0(x,y)&=E_x(x,y)\hat{\mathbf x}+E_y(x,y)\hat{\mathbf y}+E_z(x,y)\hat{\mathbf z}\\[8pt]
\mathbf B_0(x,y)&=B_x(x,y)\hat{\mathbf x}+B_y(x,y)\hat{\mathbf y}+B_z(x,y)\hat{\mathbf z}
\end{aligned}
\tag{4}
$$


Al sustituir las ecuaciones de los campos $\tilde{E}$ y $\tilde{B}$ en las ecuaciones de Maxwell, específicamente en la ley de Ampère y Faraday, se obtienen las siguientes expresiones:

$$
\begin{aligned}
\frac{\partial E_z}{\partial y}-ik_zE_y &=i\omega B_x\\[6pt]
ik_zE_x-\frac{\partial E_z}{\partial x}&=i\omega B_y\\[6pt]
\frac{\partial B_z}{\partial y}-ik_zB_y&=-i\frac{\omega}{c^2}E_x\\[6pt]
ik_zB_x-\frac{\partial B_z}{\partial x}&=-i\frac{\omega}{c^2}E_y
\end{aligned}
\tag{5}
$$

Las cuales nos dan los componentes $x$ e $y$ de los campos en función de los componentes en $z$ si se reescriben de la siguiente manera:

$$
\begin{aligned}
E_x&=\frac{i}{(\omega/c)^2-k^2} \left(k\frac{\partial E_z}{\partial x} + \omega\frac{\partial B_z}{\partial y}\right)\\[8pt]
E_y&=\frac{i}{(\omega/c)^2-k^2} \left(k\frac{\partial E_z}{\partial y} - \omega\frac{\partial B_z}{\partial x}\right)\\[8pt]
B_x&=\frac{i}{(\omega/c)^2-k^2} \left(k\frac{\partial B_z}{\partial x} - \frac{\omega}{c^2}\frac{\partial E_z}{\partial y}\right)\\[8pt]
B_y&=\frac{i}{(\omega/c)^2-k^2} \left(k\frac{\partial B_z}{\partial y} + \frac{\omega}{c^2}\frac{\partial E_z}{\partial x}\right)
\end{aligned}
\tag{6}
$$

Por lo que el problema se reduce a encontrar los componentes $E_z$ y $B_z$. Si se insertan las ecuaciones (6) en las otras 2 ecuaciones de Maxwell restantes, se obtienen las siguientes ecuaciones desacopladas:

$$
\begin{aligned}
\frac{\partial^2 E_z}{\partial x^2}
+\frac{\partial^2 E_z}{\partial y^2}
+\left[\left(\frac{\omega}{c}\right)^2-k^2\right]E_z&=0\\[10pt]
\frac{\partial^2 B_z}{\partial x^2}
+\frac{\partial^2 B_z}{\partial y^2}
+\left[\left(\frac{\omega}{c}\right)^2-k^2\right]B_z&=0
\end{aligned}
\tag{7}
$$


### Caso de analisis: Transversal Electrica (TE)

Para el caso de que la onda sea transversal eléctrica, se añade la condición de que $E_z=0$. Por lo tanto, el sistema se reduce a resolver la ecuación:

$$
\begin{aligned}
\frac{\partial^2 B_z}{\partial x^2}
+\frac{\partial^2 B_z}{\partial y^2}
+\left[\left(\frac{\omega}{c}\right)^2-k^2\right]B_z&=0
\end{aligned}
\tag{7.b}
$$

Al utilizar $B_z(x,y) = X(x)Y(y)$, se llegan a las siguientes soluciones:

$$
\begin{aligned}
X(x)&=A\cos(k_xx)+B\sin(k_xx)\\[6pt]
Y(y)&=C\cos(k_yy)+D\sin(k_yy)
\end{aligned}
\tag{8}
$$

Donde las constantes de separación deben satisfacer:

$$
k_x^2+k_y^2=\left(\frac{\omega}{c}\right)^2-k^2
\tag{9}
$$

Al aplicar las condiciones de frontera, se obtiene:

$$
B_z(x,y)=B_0
\cos\left(\frac{m\pi x}{a}\right)
\cos\left(\frac{n\pi y}{b}\right)
\tag{10}
$$

Donde $m, n = 0, 1, 2, \ldots$, pero no pueden ser ambos cero.

Además, la ecuación (9) se puede transformar en una ecuación que nos dé el valor del número de onda $k$:

$$
k=\frac{1}{c}\sqrt{\omega^2 - \omega_{mn}^2}
\tag{11}
$$

Donde $\omega_{mn}$ es la frecuencia de corte, y se define como:

$$
\omega_{mn}=c\pi\sqrt{\left(\frac{m}{a}\right)^2+\left(\frac{n}{b}\right)^2}
\tag{12}
$$

Esta frecuencia determina si el modo puede propagarse dentro de la guía:

$$
\begin{cases}
\omega > \omega_{mn} & \text{Modo propagante}\\[6pt]
\omega < \omega_{mn} & \text{Modo evanescente}
\end{cases}
$$

Para la simulación, se va a considerar el modo fundamental de propagación  $\mathrm{TE}_{10}$. Donde $m=1$ y $n=0$. Sabiendo esto, se pueden calcular los componentes de los campos electricos y magneticos:


$$
\begin{aligned}
E_x &= 0\\[4pt]
E_y &= \frac{i\omega a}{\pi}B_0\sin\left(\frac{\pi x}{a}\right)\\[4pt]
E_z &= 0\\[8pt]
B_x &= -\frac{ika}{\pi}B_0\sin\left(\frac{\pi x}{a}\right)\\[4pt]
B_y &= 0\\[4pt]
B_z &= B_0\cos\left(\frac{\pi x}{a}\right)
\end{aligned}
\tag{13}
$$

Y reescribiendo los campos con sus componentes temporales, en el caso de que $\omega > \omega_{mn}$:


$$
\begin{aligned}
E_y(x,z,t)&=-\frac{\omega a}{\pi}B_0\sin\left(\frac{\pi x}{a}\right)\sin(kz-\omega t)\\[6pt]
B_x(x,z,t)&=\frac{ka}{\pi}B_0\sin\left(\frac{\pi x}{a}\right)\sin(kz-\omega t)\\[6pt]
B_z(x,z,t)&=B_0\cos\left(\frac{\pi x}{a}\right)\cos(kz-\omega t)
\end{aligned}
\tag{14}
$$

Para el caso contrario, cuando $\omega < \omega_{mn}$, el número de onda
se vuelve imaginario, $k=i\alpha$, donde

$$
\alpha=\frac{\sqrt{\omega_{mn}^2-\omega^2}}{c}
\tag{15}
$$

En este régimen, los campos son evanescentes y presentan un decaimiento exponencial en la dirección de propagación. Por lo tanto, sus componentes temporales se pueden expresar como

$$
\begin{aligned}
E_y(x,z,t)&=-\frac{\omega a}{\pi}B_0
\sin\left(\frac{\pi x}{a}\right)e^{-\alpha z}\sin(\omega t)
\\[6pt]
B_x(x,z,t)&=-\frac{\alpha a}{\pi}B_0
\sin\left(\frac{\pi x}{a}\right)e^{-\alpha z}\cos(\omega t)
\\[6pt]
B_z(x,z,t)&=B_0
\cos\left(\frac{\pi x}{a}\right)e^{-\alpha z}\cos(\omega t)
\end{aligned}
\tag{16}
$$


## Funciones para simular el sistema

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.constants import c, pi
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib.patheffects as pe
import warnings
warnings.filterwarnings("ignore", category=RuntimeWarning)
```

```python
"""Variables del sistema"""
a = 0.04 #Ancho de la guia
b = 0.02 #Alto de la guia
l = 0.10 #Profundidad de la guia

m = 1
n = 0

B0 = 1

#Calcula la frecuencia de corte
omega_c = c*pi*np.sqrt((m/a)**2 + (n/b)**2)
```

```python
#Campo electrico en y
def Ey_field(X, Z, k, omega, t=0):
    if omega > omega_c:
        return -(omega*a/pi)*B0*np.sin(pi*X/a)*np.sin(k*Z - omega*t)
    else:
        alpha = np.sqrt(omega_c**2 - omega**2)/c
        return (omega*a/pi)*B0*np.sin(pi*X/a)*np.exp(-alpha*Z)*np.sin(omega*t)


#Campo magnetico en x
def Bx_field(X, Z, k, omega, t=0):
    if omega > omega_c:
        return (k*a/pi)*B0*np.sin(pi*X/a)*np.sin(k*Z - omega*t)
    else:
        alpha = np.sqrt(omega_c**2 - omega**2)/c
        return (alpha*a/pi)*B0*np.sin(pi*X/a)*np.exp(-alpha*Z)*np.cos(omega*t)


#Campo magnetico en z
def Bz_field(X, Z, k, omega, t=0):
    if omega > omega_c:
        return B0*np.cos(pi*X/a)*np.cos(k*Z - omega*t)
    else:
        alpha = np.sqrt(omega_c**2 - omega**2)/c
        return B0*np.cos(pi*X/a)*np.exp(-alpha*Z)*np.cos(omega*t)


def simulate(omega, z=0):
    #Calcula el numero de onda
    k = np.sqrt(omega**2 - omega_c**2)/c
    
    #Inicializa la figura
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    
    #Malla del campo transversal
    x_t = np.linspace(0.05*a, 0.95*a, 10)
    y_t = np.linspace(0.15*b, 0.85*b, 5)
    Xt, Yt = np.meshgrid(x_t, y_t)


    #Campo electrico transversal inicial
    E_max = omega*a*abs(B0)/pi
    Ex_t = np.zeros_like(Xt)
    Ey_t = Ey_field(Xt, z, k, omega) / E_max

    #Campo magnetico transversal inicial
    B_max = abs(k*a/pi*B0)
    Bx_t = Bx_field(Xt, z, k, omega) / B_max
    By_t = np.zeros_like(Xt)
    
    quiv_et = axes[0].quiver(
        Xt/a, Yt/b, Ex_t, Ey_t, Ey_t,
        cmap="RdBu",
        clim=(-1, 1),
        pivot="tail",
        angles="xy",
        scale_units="xy",
        scale=7
    )

    quiv_bt = axes[0].quiver(
        Xt/a, Yt/b, Bx_t, By_t, Bx_t,
        cmap="PiYG",
        clim=(-1, 1),
        pivot="tail",
        angles="xy",
        scale_units="xy",
        scale=10
    )

    #Malla del campo longitudinal
    z_l = np.linspace(0.01*l, 0.99*l, 19)
    x_l = np.linspace(0, a, 11)
    Zl, Xl = np.meshgrid(z_l, x_l)

    #Campo electrico longitudinal inicial
    Ey_l = Ey_field(Xl, Zl, k, omega) / E_max

    #Mapa de color del campo electrico
    mesh_el = axes[1].pcolormesh(
        Zl/l, Xl/a, Ey_l,
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        shading="auto",
        alpha=1,
        zorder=0
    )

    #Campo magnetico longitudinal inicial
    Bz_l = Bz_field(Xl, Zl, k, omega) / B0
    Bx_l = Bx_field(Xl, Zl, k, omega) / B0
    mag = np.sqrt(Bx_l**2 + Bz_l**2)

    quiv_bl = axes[1].quiver(
        Zl/l, Xl/a, Bz_l, Bx_l,
        color="black",
        pivot="mid",
        angles="xy",
        scale_units="xy",
        scale=20,
        zorder=3
    )

    quiv_bl.set_path_effects([
        pe.Stroke(linewidth=1, foreground="white"),
        pe.Normal()
    ])

    #Configura las graficas
    axes[0].set(
        xlabel=r"$x/a$",
        ylabel=r"$y/b$",
        xlim=(0, 1),
        ylim=(0, 1),
        title="Vista XY"
    )

    axes[1].set(
        xlabel=r"$z/l$",
        ylabel=r"$x/a$",
        xlim=(0, 1),
        ylim=(-0.1, 1.1),
        title="Vista XZ"
    )

    #Barra de color del campo electrico
    fig.colorbar(
        mesh_el,
        ax=axes[1],
        label=r"$E_y/E_{\max}$"
    )

    #Periodo y cantidad de fotogramas
    T = 2*pi/omega
    frames = 100

    #Actualizar el ploteo
    def update(frame):

        t = 2 * frame * T / frames

        #Actualiza el campo transversal
        Ey_t = Ey_field(Xt, z, k, omega, t) / E_max
        quiv_et.set_UVC(Ex_t, Ey_t, Ey_t)
        Bx_t = Bx_field(Xt, z, k, omega, t) / B_max
        quiv_bt.set_UVC(Bx_t, By_t, Bx_t)

        #Actualiza el campo electrico longitudinal
        Ey_l = Ey_field(Xl, Zl, k, omega, t) / E_max
        mesh_el.set_array(Ey_l.ravel())

        #Actualiza el campo magnetico longitudinal
        Bz_l = Bz_field(Xl, Zl, k, omega, t) / B0
        Bx_l = Bx_field(Xl, Zl, k, omega, t) / B0
        mag = np.sqrt(Bx_l**2 + Bz_l**2)

        quiv_bl.set_UVC(Bz_l, Bx_l)

        return quiv_et, quiv_bt, quiv_bl, mesh_el

    #Crea la animacion
    ani = FuncAnimation(fig, update, frames=frames, interval=50, blit=True)
    plt.close(fig)

    return HTML(ani.to_jshtml())

def plot_dispersion():

    #Crea un array de frecuencias mayores a la de corte
    omega = np.linspace(1.01*omega_c, 5*omega_c, 500)

    #Calcula el numero de onda
    k = np.sqrt(omega**2 - omega_c**2)/c

    #Calcula la velocidad de fase y de grupo
    vf = omega/k
    vg = c**2 * k/omega

    # Figura
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Relacion de dispersion
    axes[0].plot(k*c/omega_c, omega/omega_c)

    axes[0].set(
        xlabel=r"$ck_z/\omega_c$",
        ylabel=r"$\omega/\omega_c$",
        title="Relación de dispersión"
    )

    #Velocidades de fase y grupo
    axes[1].plot(omega/omega_c, vf/c, label=r"$v_f/c$")
    axes[1].plot(omega/omega_c, vg/c, label=r"$v_g/c$")

    axes[1].axhline(1, color="black", linestyle="--", label=r"$c$")

    axes[1].set(
        xlabel=r"$\omega/\omega_c$",
        ylabel="Velocidad normalizada",
        title="Velocidades de fase y grupo"
    )

    axes[1].set_ylim(0, 5)
    axes[1].legend()

    for ax in axes:
        ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()
```

En la siguiente parte se van a utilizar las funciones definidas para estudiar los casos de interés.

## C) Resultados e interpretación física

### ¿Por qué no cualquier frecuencia se propaga si el interior es vacío?

El motivo por el que no cualquier frecuencia puede propagarse dentro de la guía, a pesar de que su interior es vacío, es que las paredes conductoras imponen condiciones de frontera que restringen los campos electromagnéticos. En el espacio libre, una onda puede propagarse con cualquier frecuencia positiva y cumple la relación $k=\omega/c$. En cambio, dentro de la guía aparece una escala geométrica, determinada por sus dimensiones transversales, que restringe los modos de propagación posibles.

El papel de las condiciones de frontera es que, como las paredes son conductores perfectos, la componente del campo eléctrico paralela a ellas debe anularse. Para el modo $\mathrm{TE}_{10}$, esto significa que $E_y$ debe ser cero en $x=0$ y en $x=a$. Por esta razón, el perfil transversal del campo eléctrico tiene la forma $E_y\propto\sin(\pi x/a)$. 

Entonces, si $\omega>\omega_c$, el número de onda longitudinal es real y los campos oscilan en la dirección $z$, por lo que existe propagación. Por el contrario, si $\omega<\omega_c$, el número de onda longitudinal es imaginario y los campos presentan un decaimiento exponencial en esa dirección. Esto explica por qué el hecho de que el interior esté vacío no garantiza que cualquier frecuencia pueda propagarse.

### Caso $\omega<\omega_c$: modo evanescente

```python
simulate(omega=0.9*omega_c, z=0.5)
```

Al avanzar la animación, el patrón de la Vista XZ no se desplaza a lo largo de $z$: siempre ocupa la misma región, pegada a $z=0$, y se apaga hacia $z/l\approx0.5$. Lo que cambia con el tiempo es la intensidad y el signo de los campos, que se alternan. En el fotograma inicial el mapa de color de $E_y$ está en blanco y las flechas de $\mathbf B$ están en su máximo, casi horizontales ($B_z$ domina), hacia la derecha para $x<a/2$ y hacia la izquierda para $x>a/2$. Después $E_y$ aparece en rojo, más intenso en el centro del ancho y cerca de $z=0$, mientras las flechas de $\mathbf B$ se acortan; cuando $E_y$ llega a su máximo, las flechas de $\mathbf B$ casi desaparecen. Luego $E_y$ se debilita y $\mathbf B$ reaparece con las flechas invertidas. A continuación $E_y$ vuelve a crecer, ahora en azul, hasta un máximo negativo, de nuevo con $\mathbf B\approx0$, y el ciclo se repite.

Esto ocurre porque en la Ec. (16) la dependencia en $z$ y en $t$ está separada: $e^{-\alpha z}$ fija la forma del perfil, y $\sin\omega t$ y $\cos\omega t$ solo modulan su amplitud, sin ningún término $k_zz-\omega t$ que haga viajar el patrón. Como $E_y\propto\sin\omega t$ y $\mathbf B\propto\cos\omega t$ están desfasados $90^\circ$, uno es máximo cuando el otro se anula: la energía pasa de eléctrica a magnética y vuelve, sin avanzar, y por eso el promedio de $S_z$ es cero. La caída con $z$ se debe a que $k_z$ es imaginario (Ec. 15): con $\delta=1/\alpha\approx2.9$ cm, a $z=l/2$ queda cerca del 18 % de la amplitud en $z=0$.

### Caso $\omega=\omega_c$: frecuencia de corte

```python
simulate(omega=1*omega_c, z=0.5)
```

Al avanzar la animación, el patrón no se desplaza ni decae a lo largo de $z$: todas las franjas de la Vista XZ cambian a la vez, y lo único que varía es la intensidad y el signo de los campos. En el fotograma inicial no hay campo eléctrico (la Vista XY está vacía) y las flechas de $\mathbf B$ están en su máximo: son horizontales ($B_z$), idénticas a lo largo de $z$, hacia la derecha para $x<a/2$ y hacia la izquierda para $x>a/2$, más largas cerca de las paredes y nulas en $x=a/2$. Luego aparece $E_y$: en la Vista XY, flechas verticales hacia arriba, con máximo en el centro ($x/a\approx0.5$), longitud casi nula en las paredes y la misma altura en todo $y$; en la Vista XZ, franjas horizontales que se vuelven rojas, más intensas en $x=a/2$ y uniformes en $z$. Mientras $E_y$ crece, las flechas de $\mathbf B$ se acortan, y cuando $E_y$ llega a su máximo casi desaparecen. Después $E_y$ se debilita, $\mathbf B$ reaparece con los sentidos invertidos (hacia la izquierda para $x<a/2$ y hacia la derecha para $x>a/2$), $E_y$ cambia de signo (flechas hacia abajo, franjas azules) y alcanza su máximo negativo con $\mathbf B\approx0$. De ahí en adelante el ciclo se repite.

Esto ocurre porque con $\omega=\omega_c$ el número de onda longitudinal es cero (Ec. 11), así que desaparece la dependencia en $z$ y $B_x=0$ (Ec. 13). Solo quedan $E_y\propto\sin(\pi x/a)\sin\omega t$ y $B_z\propto\cos(\pi x/a)\cos\omega t$: una onda estacionaria transversal en la que cada sección de la guía oscila en el mismo instante que las demás. El desfase de $90^\circ$ entre $E_y$ y $B_z$ hace que uno sea máximo cuando el otro se anula, de modo que la energía va y viene entre forma eléctrica y magnética sin avanzar. Las flechas de $E_y$ nulas en $x=0,a$ muestran $\mathbf E_\parallel=0$ (Ec. 1), y las flechas de $\mathbf B$ solo horizontales en las paredes muestran $B_\perp=0$. Físicamente, las ondas planas que forman el modo ($\theta=90^\circ$) rebotan entre las paredes sin avanzar en $z$.

### Caso $\omega>\omega_c$: modo propagante

```python
simulate(omega=1.5*omega_c, z=0.5)
```

Al avanzar la animación, el patrón de la Vista XZ se desplaza hacia $+z$ sin cambiar de forma: en el fotograma inicial se ven tres bandas de $E_y$ (azul, roja, azul) separadas unos $0.7\,l$, y poco después entra una banda roja por $z=0$ mientras las demás avanzan hacia la derecha. En medio periodo el patrón queda con los colores invertidos, es decir, desplazado media longitud de onda ($\approx0.36\,l$), y al completar el periodo vuelve a ser idéntico al inicial. Todas las bandas son más intensas en el centro del ancho ($x=a/2$) y se anulan en las paredes. Las flechas de $\mathbf B$ son verticales ($B_x$) en el centro de cada banda, donde $E_y$ es máximo, y horizontales ($B_z$) entre bandas y junto a las paredes, donde alternan de sentido a lo largo de $z$; en conjunto forman lazos que viajan con las bandas.

En la Vista XY, las flechas de $E_y$ (verticales) y de $B_x$ (horizontales) crecen y se anulan al mismo tiempo, con máximo en el centro y longitud casi nula en las paredes. Cuando $E_y$ apunta hacia arriba, $B_x$ apunta hacia la izquierda, y al invertirse una se invierte la otra, de modo que $\mathbf E\times\mathbf B$ siempre apunta hacia $+z$. Esto contrasta con el régimen evanescente, donde $E_y$ y $\mathbf B$ nunca eran máximos a la vez.

Esto ocurre porque en la Ec. (14) los campos dependen de $k_zz-\omega t$, por lo que el patrón se traslada hacia $+z$ y se repite cada $\lambda_g=2\pi/k_z\approx0.72\,l$ (Ec. 11). $E_y$ y $B_x$ comparten la misma fase, con signos opuestos, y por eso $\langle S_z\rangle>0$: hay flujo neto de energía. $B_z$ está desfasado $90^\circ$ en $z$, y por eso es máximo entre bandas y vale cero en su centro. Las flechas horizontales junto a $x=0$ y $x=a$ muestran $B_\perp=0$ y el valor nulo de $E_y$ en las paredes muestra $E_\parallel=0$ (Ec. 1). El patrón de fase avanza a $v_f\approx1.34\,c$, mientras que la energía lo hace a $v_g\approx0.75\,c$ (ver figura de velocidades).

### ¿Por qué $v_f>c$ no implica una transmisión superlumínica de energía o información?

```python
plot_dispersion()
```

Que $v_f>c$ no implica transmisión superlumínica de energía o información. En el panel derecho de la gráfica, $v_f/c$ (azul) diverge cuando $\omega\to\omega_c^+$ y baja hacia $1$ al aumentar la frecuencia, mientras que $v_g/c$ (naranja) parte de casi cero en $\omega_c$ y sube hacia $1$ sin superar nunca la línea punteada de $c$. Las dos curvas son recíprocas, ya que $v_fv_g=c^2$: por ejemplo, para $\omega=1.5\,\omega_c$ se lee $v_f\approx1.34\,c$ y $v_g\approx0.75\,c$. En el panel izquierdo ocurre lo mismo de otra forma: la curva $\omega(k_z)$ queda por encima de la recta $\omega=ck_z$, y por eso $v_f=\omega/k_z>c$, pero su pendiente local, que es $v_g=d\omega/dk_z$, es menor que $c$, y casi nula cerca del corte.

La velocidad de fase describe cómo se desplaza una cresta de una onda monocromática ideal, que no tiene principio ni fin y por tanto no transporta información. Geométricamente, las dos ondas planas del modo avanzan en zigzag a velocidad $c$, y el punto donde sus crestas cruzan el eje $z$ se mueve más rápido que $c$, como el cruce de unas tijeras que se cierran. La energía y la información, en cambio, viajan con la velocidad de grupo, que siempre es menor que $c$.


## D) Retrospectiva

Lo que mejor comprendí con la visualización fue cómo cambia el comportamiento de los campos al cruzar $\omega_c$, tanto en $z$ como en el tiempo. Con $\omega=0.9\,\omega_c$ los campos decaen con $z$ (Ec. 16) y oscilan en su lugar; con $\omega=\omega_c$ no dependen de $z$ y toda la guía oscila a la vez; con $\omega=1.5\,\omega_c$ forman bandas que viajan hacia $+z$ (Ec. 14). Al pasar de $1.5\,\omega_c$ a $\omega_c$, las bandas se ensanchan ($\lambda_g\approx0.72\,l$ en el primer caso y $\lambda_g\to\infty$ en el segundo) hasta desaparecer, y todos los puntos quedan oscilando en fase. Eso es lo que significa que $v_f\to\infty$ cuando $\omega\to\omega_c^+$: el patrón de fase deja de variar en $z$, pero esto no implica transporte de energía a esa velocidad, ya que $v_g\to0$. Además, las flechas de $\mathbf B$, que para $1.5\,\omega_c$ son verticales en el centro de las bandas y horizontales entre ellas, se vuelven todas horizontales en $\omega_c$, pues el cociente de amplitudes $B_x/B_z=k_za/\pi$ tiende a cero. Una limitación es que la onda es monocromática: $v_f$ se observa directamente, pero $v_g$ solo se infiere de la relación de dispersión.

También comprendí mejor la relación temporal entre $E_y$ y $\mathbf B$. Con $\omega=0.9\,\omega_c$ y con $\omega=\omega_c$, $E_y$ es máximo cuando $\mathbf B$ se anula y viceversa, porque $E_y\propto\sin\omega t$ y $\mathbf B\propto\cos\omega t$: los campos se turnan y el patrón oscila en su lugar. Con $1.5\,\omega_c$, en cambio, $E_y$ y $B_x$ crecen y se anulan a la vez, y el patrón completo se traslada. El factor $e^{-\alpha z}$ de la Ec. (16) podría sugerir una onda que se propaga mientras se amortigua, pero la animación muestra que no: la amplitud cae con $z$ (cerca del 18 % de la amplitud en $z=0$ a $z=l/2$, consistente con $\delta\approx2.9$ cm) sin que nada se desplace. Esta diferencia corresponde al promedio de $S_z$: nulo cuando los campos se turnan, positivo cuando coinciden. Así, la visualización permite distinguir comportamientos que las ecuaciones describen pero que no siempre resultan intuitivos.
